```python
import torch.nn as nn
import torch as torch
class SelfAttention(nn.Module):
  def __init__(self, embed_size):
    super().__init__()
    self.embed_size = embed_size
    self.q_w = nn.Linear(embed_size, embed_size, bias=False)
    self.k_w = nn.Linear(embed_size, embed_size, bias=False)
    self.v_w = nn.Linear(embed_size, embed_size, bias=False)
  def forward(self, token_embeddings):
    q = self.q_w(token_embeddings)
    k = self.k_w(token_embeddings)
    v = self.v_w(token_embeddings)
    sims = q @ k.transpose(-2, -1)
    scaled_sims = sims / k.size(-1)
    attention_percents = torch.softmax(scaled_sims, -1)
    output = attention_percents @ v
self_attention = SelfAttention(2)
token_embeddings = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
attention_scores = self_attention(token_embeddings)
print(attention_scores)
```
